```python
def clean_timestamp(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # Unix timestamp
    if value.isdigit():
        return pd.to_datetime(int(value), unit="s")

    # Normal datetime / DD/MM/YYYY datetime
    return pd.to_datetime(value, dayfirst=True, errors="coerce")
```

```python
outages["start_ts"] = outages["start_ts"].apply(clean_timestamp)
outages["end_ts"] = outages["end_ts"].apply(clean_timestamp)
```

```python
def clean_duration(value):
  if pd.isna(value):
    return pd.NA
  value = str(value).strip().lower()
  value = value.replace("min","").strip()

  return pd.to_numeric(value, errors="coerce")
```
